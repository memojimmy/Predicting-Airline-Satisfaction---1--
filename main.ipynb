## numerical Pipeline

```python
from feature_engine.outliers import Winsorizer

outlier_handler= Winsorizer(capping_method ="iqr",
                             tail ="both", fold = 1.5,
                               variables = ["Flight Distance" ])
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

scaler = StandardScaler()

scaler_pipeline = Pipeline([
    ('outlier_handler', outlier_handler),
    ('scaler', StandardScaler())
])
scaler_pipeline 
```

## Categorical Pipeline

```python
x_cat = x.select_dtypes(include='str')
x_cat
```

```python
ord_col = ['Class']
ord_col
```

```python
# Ordinal Encoding 

from sklearn.preprocessing import OrdinalEncoder

ord_encoder = OrdinalEncoder()

ord_encoder_pipe = Pipeline([
    ('ord_encoder', ord_encoder)
])
ord_encoder_pipe
```

```python
onehot_cols = x.select_dtypes('str').columns.drop('Class')
onehot_cols
```

```python
# One Hot Encoding 
from sklearn.preprocessing import OneHotEncoder

onehot_encoder = OneHotEncoder(sparse_output=False, drop='first')
onehot_encoder_pipe = Pipeline([
    ('onehot_encoder', onehot_encoder)
])
onehot_encoder_pipe
```

```python
y.value_counts()
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ('scaler', scaler_pipeline, x_numerical.columns),
    ('ord_encoder', ord_encoder_pipe, ord_col),
    ('onehot_encoder', onehot_encoder_pipe, onehot_cols)
])  
preprocessor
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_validate 

model_pipeline = Pipeline([ ('preprocessor', preprocessor) ,('logistic_regression', LogisticRegression())  ])

cv = cross_validate(estimator= model_pipeline, X=x, y=y, cv=5, scoring='accuracy' , return_train_score=True , n_jobs=-1 )

print(f"Train Accuracy: {cv['train_score'].round(3).mean()*100} \nTest Accuracy: {cv['test_score'].round(3).mean()*100}")
```

```python
from sklearn.model_selection import cross_validate 
from sklearn.linear_model import LogisticRegression 
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC 
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier 
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
```

```python
models = [('LogisticRegression',LogisticRegression(random_state=42 , n_jobs=-1)),
          ('KNeighborsClassifier',KNeighborsClassifier()),
          ('GaussianNB',GaussianNB()),
        #   ('SVC',SVC(kernel='rbf',random_state=42)),
        #   ('SVC',SVC(kernel='linear',random_state=42)),
          ('DecisionTreeClassifier',DecisionTreeClassifier(random_state=42,)),
          ('RandomForestClassifier',RandomForestClassifier(random_state=42,n_jobs=-1)),
          ('XGBClassifier',XGBClassifier(random_state=42, n_jobs=-1)),
          ('CatBoostClassifier',CatBoostClassifier(verbose=0,random_state=42)),
          ('LGBMClassifier',LGBMClassifier(random_state=42, n_jobs=-1, verbose=-1))

          ]

for model in models:
    model_pipeline = Pipeline(steps=[('preprocessing' , preprocessor),
                                     ('model' , model[1])])
    cv = cross_validate(model_pipeline , X=x , y=y , cv=5 , scoring='accuracy' , n_jobs=-1 , return_train_score=True )
    print(f"Train Accuracy: {cv['train_score'].round(3).mean()*100} \nTest Accuracy: {cv['test_score'].round(3).mean()*100}")
    print('=' * 50)
```

```python
model_pipeline = Pipeline([ ('preprocessor', preprocessor) ,('random forest', RandomForestClassifier(random_state=42 ,n_estimators=300 , max_depth=8 ,n_jobs=-1))])

cv = cross_validate(estimator= model_pipeline, X=x, y=y, cv=5, scoring='accuracy' , return_train_score=True , n_jobs=-1 )

print(f"Train Accuracy: {cv['train_score'].round(3).mean()*100} \nTest Accuracy: {cv['test_score'].round(3).mean()*100}")
```

```python
model_pipeline = Pipeline([ ('preprocessor', preprocessor) ,('Xgbclassifier', XGBClassifier(random_state=42 ,n_estimators=1000,learning_rate=0.05,max_depth=5,))])

cv = cross_validate(estimator= model_pipeline, X=x, y=y, cv=5, scoring='accuracy' , return_train_score=True , n_jobs=-1 )

print(f"Train Accuracy: {cv['train_score'].round(3).mean()*100} \nTest Accuracy: {cv['test_score'].round(3).mean()*100}")
```
